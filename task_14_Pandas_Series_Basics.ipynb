```python
import pandas as pd

# 1. Create a Student Marks Series
marks = pd.Series(
    [85, 92, 78, 88, 95],
    index=["Rahul", "Priya", "Amit", "Sneha", "Kiran"],
    name="Marks"
)

# 2. Display all student marks
print("STUDENT MARKS")
print(marks)

# 3. Access individual student marks
print("\nPriya's marks:")
print(marks.loc["Priya"])

# 4. Filter students scoring above 85
print("\nStudents scoring above 85:")
print(marks[marks > 85])

# 5. Perform arithmetic
print("\nMarks after adding 5 bonus points:")
print(marks + 5)

# 6. Calculate statistics
print("\nSTATISTICAL ANALYSIS")
print("Total marks:", marks.sum())
print("Average marks:", marks.mean())
print("Highest marks:", marks.max())
print("Lowest marks:", marks.min())
print("Median marks:", marks.median())
print("Number of students:", marks.count())

# 7. Display the descriptive summary
print("\nDESCRIPTIVE SUMMARY")
print(marks.describe())
```

```python
import pandas as pd

sales = pd.Series(
    [1200, 1500, 900, 1800, 2100],
    index=["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"],
    name="Daily Sales"
)

print("Daily Sales:")
print(sales)

print("\nTotal sales:", sales.sum())
print("Average daily sales:", sales.mean())
print("Highest sales:", sales.max())
print("Days with sales above 1000:")
print(sales[sales > 1000])

print("\nSales summary:")
print(sales.describe())
```
